```python
import json

# Lire les 25 359 restaurants (une ligne JSON par restaurant)
with open("/donnees/mongo/restaurants.json", encoding="utf-8") as fichier:
    restaurants = [json.loads(ligne) for ligne in fichier]

def serveur_par_hachage(restaurant, nb_serveurs):
    # empreinte : les deux derniers chiffres du numéro
    empreinte = int(restaurant["restaurant_id"][-2:])
    return empreinte % nb_serveurs

for nb_serveurs in (3, 4):
    charge = [0] * nb_serveurs
    for restaurant in restaurants:
        charge[serveur_par_hachage(restaurant, nb_serveurs)] += 1
    print(nb_serveurs, "serveurs :", charge)

nb_deplaces = 0
for restaurant in restaurants:
    if serveur_par_hachage(restaurant, 3) != serveur_par_hachage(restaurant, 4):
        nb_deplaces += 1
print("Changent de serveur :", nb_deplaces, "sur", len(restaurants))
```
